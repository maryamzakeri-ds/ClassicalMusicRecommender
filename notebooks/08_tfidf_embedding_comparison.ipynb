Several musical works are used as queries and the recommended results from both models are compared to evaluate similarities and differences.

## Model Comparison Conclusion

Because the embedding model was built from the same structured metadata used by the TF-IDF baseline, both approaches produced very similar recommendations.
This suggests that the quality of the representation is limited by the available metadata rather than the similarity algorithm itself.
