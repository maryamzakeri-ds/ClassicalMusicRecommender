# Data Cleaning

## Objective

The goal of this notebook is to clean and prepare raw dataset collected from the Open Opus API.

The following steps will be performed:
- Remove features that do not provide useful information for the recommendation system.
- Remove redundant composer name information.
- Convert composer birth and death dates from full date format into year format.

```python
df_clean = df.drop(['portrait', 'searchterms', 'name'], axis=1)
```

```python
df_clean = df_clean.rename(columns={'complete_name': 'composer_name'})
```

```python
df_clean.columns
```

```python
df_clean['birth'] = pd.to_datetime(df_clean['birth']).dt.year
df_clean['death'] = pd.to_datetime(df_clean['death']).dt.year
```

```python
df_clean[['birth', 'death']].head()
```

```python
df_clean.isna().sum()
```

### Save Cleaned Dataset

```python
df_clean.to_csv(Path('../data/processed/works_with_composers_cleaned.csv'), index=False)
```

## Data Cleaning Summary

In this section, the raw dataset was cleaned and prepared for the next stages of the project.

The following changes were applied:
- Removed unnecessary features such as portrait, search terms and redundant composer name information.
- Transformed composer birth and death dates from full date format into year format.
- Noticed that death year is NaN for some composers. It might mean they are still alive. They will stay NaN for further analysis.
- The subtitle column also has several null values, but it may be useful for feature engineering to extract new features from the combination of title and subtitle.
